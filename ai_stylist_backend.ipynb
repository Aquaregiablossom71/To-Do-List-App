```python
def estimate_body_shape(height, weight, gender):
    """Return a heuristic body shape based on BMI and gender."""
    bmi = weight / ((height / 100) ** 2)
    if gender.lower() == 'female':
        if bmi < 18.5:
            return 'rectangle'
        elif bmi < 24:
            return 'pear'
        elif bmi < 28:
            return 'hourglass'
        else:
            return 'apple'
    else:
        if bmi < 20:
            return 'rectangle'
        elif bmi < 25:
            return 'triangle'
        elif bmi < 30:
            return 'inverted'
        else:
            return 'oval'
```

```python
# Add body shape column
df['body_shape'] = df.apply(lambda row: estimate_body_shape(row['height_cm'], row['weight_kg'], row['gender']), axis=1)
```

```python
def time_color_hint(time_of_day):
    """Simple hint for color palettes by time."""
    t = time_of_day.lower()
    if t in ['morning', 'day']:
        return "lighter / pastel tones recommended"
    if t == 'evening':
        return "vibrant / bright tones recommended"
    if t == 'night':
        return "darker / richer tones recommended"
    return "neutral tones recommended"
```

```python
def recommend_outfits(gender, height, weight, skin_tone, event_type,
                       dress_code=None, time_of_day=None, style_preference=None, top_n=3):
    """
    Score dataset entries on multiple factors and return top_n recommendations.
    Scoring components:
      - exact match on gender (+2)
      - event_type match (+3)
      - dress_code match (+2)
      - time_of_day match (+1)
      - style_preference match (+2)
      - skin_tone match (+1)
      - body_shape match (+2)
    """
    user_body_shape = estimate_body_shape(height, weight, gender)
    print(f"\n🧠 Estimated body shape: {user_body_shape}  (BMI heuristic)")

    candidates = df.copy()

    # Basic filter: same gender (required)
    candidates = candidates[candidates['gender'].str.lower() == gender.lower()].copy()

    # Score calculation
    def score_row(row):
        score = 0
        # event
        if row['event_type'].lower() == event_type.lower():
            score += 3
        # dress code
        if dress_code and row['dress_code'].lower() == dress_code.lower():
            score += 2
        # time
        if time_of_day and row['time_of_day'].lower() == time_of_day.lower():
            score += 1
        # style
        if style_preference and row['style_preference'].lower() == style_preference.lower():
            score += 2
        # skin tone
        if row['skin_tone'].lower() == skin_tone.lower():
            score += 1
        # body shape
        if row['body_shape'].lower() == user_body_shape.lower():
            score += 2
        return score

    candidates['score'] = candidates.apply(score_row, axis=1)

    # If nothing sufficiently matched, relax filters (ensure we always return something)
    candidates = candidates.sort_values(by='score', ascending=False)
    top = candidates.head(top_n)

    # If top results have score 0, fallback to gender-only random sample
    if top['score'].max() == 0:
        fallback = df[df['gender'].str.lower() == gender.lower()].sample(n=min(top_n, len(df)))
        top = fallback

    return top
```

```python
# ai_stylist_backend.py
import pandas as pd
import numpy as np
import pickle

# Load dataset
df = pd.read_csv("ai_stylist_full_dataset.csv")

# --------------- Functions ---------------
def estimate_body_shape(height, weight, gender):
    bmi = weight / ((height / 100) ** 2)
    if gender.lower() == 'female':
        if bmi < 18.5:
            return 'rectangle'
        elif bmi < 24:
            return 'pear'
        elif bmi < 28:
            return 'hourglass'
        else:
            return 'apple'
    else:
        if bmi < 20:
            return 'rectangle'
        elif bmi < 25:
            return 'triangle'
        elif bmi < 30:
            return 'inverted'
        else:
            return 'oval'

def time_color_hint(time_of_day):
    t = time_of_day.lower()
    if t in ['morning', 'day']:
        return "lighter / pastel tones recommended"
    if t == 'evening':
        return "vibrant / bright tones recommended"
    if t == 'night':
        return "darker / richer tones recommended"
    return "neutral tones recommended"

def recommend_outfits(gender, height, weight, skin_tone, event_type,
                       dress_code=None, time_of_day=None, style_preference=None, top_n=3):
    user_body_shape = estimate_body_shape(height, weight, gender)
    candidates = df.copy()
    candidates = candidates[candidates['gender'].str.lower() == gender.lower()].copy()

    def score_row(row):
        score = 0
        if row['event_type'].lower() == event_type.lower():
            score += 3
        if dress_code and row['dress_code'].lower() == dress_code.lower():
            score += 2
        if time_of_day and row['time_of_day'].lower() == time_of_day.lower():
            score += 1
        if style_preference and row['style_preference'].lower() == style_preference.lower():
            score += 2
        if row['skin_tone'].lower() == skin_tone.lower():
            score += 1
        if row['body_shape'].lower() == user_body_shape.lower():
            score += 2
        return score

    candidates['score'] = candidates.apply(score_row, axis=1)
    candidates = candidates.sort_values(by='score', ascending=False)
    top = candidates.head(top_n)
    if top['score'].max() == 0:
        top = df[df['gender'].str.lower() == gender.lower()].sample(n=min(top_n, len(df)))
    return top

# --------------- Save as Pickle ---------------
model_data = {
    "df": df,
    "estimate_body_shape": estimate_body_shape,
    "time_color_hint": time_color_hint,
    "recommend_outfits": recommend_outfits
}

with open("ai_stylist_model.pkl", "wb") as f:
    pickle.dump(model_data, f)

print("✅ AI Stylist model successfully saved as ai_stylist_model.pkl")
```
